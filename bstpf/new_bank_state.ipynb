```python
def get_Tables(file_path, flav):
    tables = cam.read_pdf(filepath=file_path, pages=pgs,flavor=flav)
    df = pd.DataFrame()

    for table in tables:
        df = pd.concat([df, table.df], ignore_index=True)

    return df
```

```python
unwanted_patterns = [
    'Transaction Details Page', 
    '-----------------', 
    'Page Total', 
    'Note: Cheques received',
    'returning on the basis',
    'Unless the constituent',
    'BANK OF BARODA',
    'PALI ROAD, BANDRA',
    'ADDRESS:',
    'HELPLINE NO.',
    'BRANCH PHONE NO.',
    'MICR CODE:',
    'A/C Number',
    'Statement of account',
    'DATE PARTICULARS'
]

# 3. Create a single regex pattern by joining the list with '|' (which means OR)
# This finds rows that contain 'pattern1' OR 'pattern2' OR 'pattern3', etc.
regex_pattern = '|'.join(unwanted_patterns)

# 4. Filter the DataFrame
# The `~` symbol inverts the selection, so we KEEP rows that DO NOT contain the patterns.
# `na=False` ensures that empty cells don't cause an error.
cleaned_df = df_main[~df_main[0].str.contains(regex_pattern, na=False)]


print("\n\n----------- DataFrame After Cleaning -----------")
print(cleaned_df)
```

```python
cleaned_df.drop(1, axis=1, inplace=True)
```

```python
import re
```

```python
def clean_currency(value):
    """Removes commas, 'Cr', and converts to a float."""
    if value is None:
        return None
    # Remove all non-digit characters except for the decimal point
    cleaned_value = re.sub(r'[^\d.]', '', str(value))
    if cleaned_value:
        return float(cleaned_value)
    return None
```

```python
processed_rows = []
```

```python
# 2. Variable to keep track of the last balance
previous_balance_numeric = None
i = 0

while i < len(cleaned_df):
    current_line = re.sub(r'^\d+\s+', '', cleaned_df[0].iloc[i])

    if re.match(r'^\d{2}-\d{2}-\d{2}', current_line):
        full_transaction_line = current_line
        
        # Merge logic (remains the same)
        if (i + 1) < len(cleaned_df):
            next_line = re.sub(r'^\d+\s+', '', cleaned_df[0].iloc[i + 1])
            if not re.match(r'^\d{2}-\d{2}-\d{2}', next_line):
                parts = current_line.split()
                description_part1 = ' '.join(parts[1:-2])
                full_transaction_line = f"{' '.join(parts[:-2])} {next_line} {' '.join(parts[-2:])}"
                i += 1 # Manually increment counter since we consumed the next line

        # Parsing logic for the (potentially merged) line
        parts = full_transaction_line.split()
        date = parts[0]
        balance_str = parts[-1]
        
        withdrawal = np.nan # Use numpy's "Not a Number" for blank cells
        deposit = np.nan
        
        # Handle the "Brought Forward" row
        if "B/F" in parts:
            description = 'B/F'
            previous_balance_numeric = clean_currency(balance_str)
        else:
            amount_str = parts[-2]
            description = ' '.join(parts[1:-2])
            
            amount_numeric = clean_currency(amount_str)
            current_balance_numeric = clean_currency(balance_str)

            # 3. Compare balances if a previous balance exists
            if previous_balance_numeric is not None and current_balance_numeric is not None:
                if current_balance_numeric < previous_balance_numeric:
                    withdrawal = amount_numeric # It's a withdrawal
                else:
                    deposit = amount_numeric # It's a deposit
            
            # 4. Update the tracker for the next iteration
            previous_balance_numeric = current_balance_numeric

        processed_rows.append({
            'Date': date,
            'Description': description,
            'Withdrawal': withdrawal,
            'Deposit': deposit,
            'Balance': balance_str
        })
        i += 1
    else:
        i += 1

# Create the final DataFrame
result_df = pd.DataFrame(processed_rows)

print(result_df)
```

```python
df_saras.replace('', np.nan, inplace=True)
```

```python
processed_data = []

for i in range(len(df_saras)):
    current_row = df_saras.iloc[i]

    # Find the anchor row (the one with the date)
    if pd.notna(current_row[0]):
        
        description_parts = []

        # Step 1: Get description from the row ABOVE
        # Check if row exists AND its date column is empty
        if i > 0:
            row_above = df_saras.iloc[i-1]
            if pd.isna(row_above[0]) and pd.notna(row_above[1]):
                description_parts.append(str(row_above[1]))

        # Step 2: Get description from the ANCHOR row itself
        if pd.notna(current_row[1]):
            description_parts.append(str(current_row[1]))

        # Step 3: Get description from the row BELOW
        # Check if row exists AND its date column is empty
        if i < len(df_saras) - 1:
            row_below = df_saras.iloc[i+1]
            if pd.isna(row_below[0]) and pd.notna(row_below[1]):
                description_parts.append(str(row_below[1]))
        
        # Combine the valid description parts
        full_description = ' '.join(description_parts)

        # Assemble the final record
        record = {
            'Date': current_row[0],
            'Description': full_description,
            'Cheque No': current_row[2],
            'Withdrawal': current_row[3],
            'Deposit': current_row[4],
            'Balance': current_row[5]
        }
        processed_data.append(record)

# Create the final DataFrame
result_df = pd.DataFrame(processed_data)

print(result_df)
```

```python
df_hdfc2.replace('', np.nan, inplace=True)
```

```python
df_hdfc2.columns = ["Date", "Description", "Chq No", "Value Date", "Withdraw", "Deposit", "Closing"]
```

```python
df_hdfc2['block_id'] = df_hdfc2['Date'].notna().cumsum()

# 3. Define a function to consolidate each block into a single row.
def consolidate_block(group):
    # The first row of the group contains the main transaction data.
    anchor_row = group.iloc[0]
    
    # Join all non-empty description parts from the entire group.
    full_description = ' '.join(group['Description'].dropna().astype(str))
    
    # Return a new pandas Series, which will become a row in our new DataFrame.
    return pd.Series({
        'Date': anchor_row['Date'],
        'Description': full_description,
        'Value Date': anchor_row['Value Date'],
        'Withdraw': anchor_row['Withdraw'],
        'Deposit': anchor_row['Deposit'],
        'Closing': anchor_row['Closing']

    })

# 4. Group by the block_id and apply our consolidation function.
# .reset_index(drop=True) cleans up the output DataFrame.
HDFC2_result_df = df_hdfc2.groupby('block_id').apply(consolidate_block).reset_index(drop=True)
```

```python
# Helper function to convert currency strings to a number
def clean_currency(value):
    if pd.isna(value): return None
    cleaned_value = re.sub(r'[^\d.]', '', str(value))
    return float(cleaned_value) if cleaned_value else None

processed_data = []
previous_balance = None

for index, line in enumerate(df_wb2[0]):
    try:
        if not line or line.isspace(): continue
        parts = line.split()

        # Step 1: Check for the "Brought Forward" special case with the NEW format
        if parts[0] == "Brought" and parts[1] == "Forward":
            # This line format is: ['Brought', 'Forward', ':', amount, balance]
            # The important value is the final balance at index 4.
            balance_amt_str = parts[4]
            record = {
                'Date': None, 'Tran': None, 'Ref Num': None,
                'Particulars': 'Brought Forward',
                'Debit Amt.': np.nan,   # No debit on a B/F line
                'Credit Amt.': np.nan,  # No credit on a B/F line
                'Balance Amt.': balance_amt_str,
                'Contradate': np.nan
            }
            # Set the initial balance for the debit/credit logic to work on the next line
            previous_balance = clean_currency(balance_amt_str)
            processed_data.append(record)
            continue # Go to the next line

        # --- Standard transaction parsing logic (remains the same) ---
        date = parts[0][:10]
        tran = parts[0][10:]
        balance_amt_str = parts[-1]
        amount_str = parts[-2]
        
        description_block = parts[1:-2]
        
        ref_num = None
        if description_block and description_block[0].isdigit() and len(description_block[0]) > 6:
            ref_num = description_block.pop(0)
        
        particulars = ' '.join(description_block)
        
        debit_amt, credit_amt = np.nan, np.nan
        current_balance = clean_currency(balance_amt_str)
        amount_numeric = clean_currency(amount_str)
        if previous_balance is not None and current_balance is not None:
            if current_balance < previous_balance:
                debit_amt = amount_numeric
            else:
                credit_amt = amount_numeric
        previous_balance = current_balance

        processed_data.append({
            'Date': date, 'Tran': tran, 'Ref Num': ref_num,
            'Particulars': particulars, 'Debit Amt.': debit_amt,
            'Credit Amt.': credit_amt, 'Balance Amt.': balance_amt_str,
            'Contradate': np.nan
        })
        
    except IndexError:
        print(f"WARNING: Skipping badly formatted line at original index {index}. Line content: '{line}'")
        continue
    except Exception as e:
        print(f"WARNING: An unexpected error occurred at original index {index}. Line: '{line}'. Error: {e}")
        continue

result_df = pd.DataFrame(processed_data)

print(result_df)
```
