```python
import camelot as cam
import pandas as pd
import numpy as np
```

```python
pgs = "all"
```

```python
def get_Tables(file_path, flav):
    tables = cam.read_pdf(filepath=file_path, pages=pgs,flavor=flav)
    df = pd.DataFrame()

    for table in tables:
        df = pd.concat([df, table.df], ignore_index=True)

    return df
```

```python
for index, row in df_main.iterrows():
    # 4. Collect data for a transaction block
    if pd.notna(row['Date']):
        temp_date = row['Date']
    if pd.notna(row['Withdrawal']):
        temp_withdrawal = row['Withdrawal']
    if pd.notna(row["Deposits"]):
        temp_deposit = row['Deposits']
    if pd.notna(row['Balance']):
        temp_balance = row['Balance']

    Particulars_cell = row['Particulars']
    if pd.notna(Particulars_cell):
        # 5. Check if the line marks the end of a transaction
        if Particulars_cell.strip().startswith('Chq:'):
            # Extract the cheque number
            cheque_no = Particulars_cell.replace('Chq:', '').strip()
            
            # Combine the description parts into a single string
            full_Particulars = ' '.join(temp_Particulars_parts)
            
            # Assemble the complete record
            processed_data.append({
                'Date': temp_date,
                'Particulars': full_Particulars,
                'Cheque Number': cheque_no,
                'Deposits': temp_deposit,
                'Withdrawal': temp_withdrawal,
                'Balance': temp_balance
            })
            
            # 6. Reset temporary variables for the next transaction
            temp_date = None
            temp_Particulars_parts = []
            temp_withdrawal = None
            temp_balance = None
        else:
            # If not a 'Chq:' line, add it to the description
            temp_Particulars_parts.append(Particulars_cell)
```

BANK OF BARODA 1

```python
df_main = get_Tables("bs/BANK OF BARODA-61-82.pdf", flav="stream")
```

```python
unwanted_patterns = [
    'Transaction Details Page', 
    '-----------------', 
    'Page Total', 
    'Note: Cheques received',
    'returning on the basis',
    'Unless the constituent',
    'BANK OF BARODA',
    'PALI ROAD, BANDRA',
    'ADDRESS:',
    'HELPLINE NO.',
    'BRANCH PHONE NO.',
    'MICR CODE:',
    'A/C Number',
    'Statement of account',
    'DATE PARTICULARS'
]

# 3. Create a single regex pattern by joining the list with '|' (which means OR)
# This finds rows that contain 'pattern1' OR 'pattern2' OR 'pattern3', etc.
regex_pattern = '|'.join(unwanted_patterns)

# 4. Filter the DataFrame
# The `~` symbol inverts the selection, so we KEEP rows that DO NOT contain the patterns.
# `na=False` ensures that empty cells don't cause an error.
cleaned_df = df_main[~df_main[0].str.contains(regex_pattern, na=False)]


print("\n\n----------- DataFrame After Cleaning -----------")
print(cleaned_df)
```

```python
cleaned_df.drop(1, axis=1, inplace=True)
```

```python
import re
```

```python
def clean_currency(value):
    """Removes commas, 'Cr', and converts to a float."""
    if value is None:
        return None
    # Remove all non-digit characters except for the decimal point
    cleaned_value = re.sub(r'[^\d.]', '', str(value))
    if cleaned_value:
        return float(cleaned_value)
    return None
```

```python
processed_rows = []
# 2. Variable to keep track of the last balance
previous_balance_numeric = None
i = 0

while i < len(cleaned_df):
    current_line = re.sub(r'^\d+\s+', '', cleaned_df[0].iloc[i])

    if re.match(r'^\d{2}-\d{2}-\d{2}', current_line):
        full_transaction_line = current_line
        
        # Merge logic (remains the same)
        if (i + 1) < len(cleaned_df):
            next_line = re.sub(r'^\d+\s+', '', cleaned_df[0].iloc[i + 1])
            if not re.match(r'^\d{2}-\d{2}-\d{2}', next_line):
                parts = current_line.split()
                description_part1 = ' '.join(parts[1:-2])
                full_transaction_line = f"{' '.join(parts[:-2])} {next_line} {' '.join(parts[-2:])}"
                i += 1 # Manually increment counter since we consumed the next line

        # Parsing logic for the (potentially merged) line
        parts = full_transaction_line.split()
        date = parts[0]
        balance_str = parts[-1]
        
        withdrawal = np.nan # Use numpy's "Not a Number" for blank cells
        deposit = np.nan
        
        # Handle the "Brought Forward" row
        if "B/F" in parts:
            description = 'B/F'
            previous_balance_numeric = clean_currency(balance_str)
        else:
            amount_str = parts[-2]
            description = ' '.join(parts[1:-2])
            
            amount_numeric = clean_currency(amount_str)
            current_balance_numeric = clean_currency(balance_str)

            # 3. Compare balances if a previous balance exists
            if previous_balance_numeric is not None and current_balance_numeric is not None:
                if current_balance_numeric < previous_balance_numeric:
                    withdrawal = amount_numeric # It's a withdrawal
                else:
                    deposit = amount_numeric # It's a deposit
            
            # 4. Update the tracker for the next iteration
            previous_balance_numeric = current_balance_numeric

        processed_rows.append({
            'Date': date,
            'Description': description,
            'Withdrawal': withdrawal,
            'Deposit': deposit,
            'Balance': balance_str
        })
        i += 1
    else:
        i += 1

# Create the final DataFrame
result_df = pd.DataFrame(processed_rows)

print(result_df)
```

```python
result_df.to_excel("BANK OF BARODA-61-82.xlsx", engine="xlsxwriter")
```
