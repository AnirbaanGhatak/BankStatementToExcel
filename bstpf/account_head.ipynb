```python
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB

def model_creation(labeled_df, unlabeled_df):
    labeled_df['Narration'] = labeled_df['Narration'].fillna('')
    unlabeled_df['Narration'] = unlabeled_df['Narration'].fillna('')

    labeled_df['Account Head'] = labeled_df['Account Head'].replace(r'^\s*$', np.nan, regex=True)
    unlabeled_df['Account Head'] = unlabeled_df['Account Head'].replace(r'^\s*$', np.nan, regex=True)

    X_train = labeled_df['Narration']
    y_train = labeled_df['Account Head']
    X_predict = unlabeled_df['Narration']

    model = Pipeline([
        ('vectorizer', TfidfVectorizer(stop_words='english', ngram_range=(1, 2))),
        ('classifier', MultinomialNB(alpha=0.1)) ])
    
    model.fit(X_train, y_train)

    return model
```

```python
model = Pipeline([
        ('vectorizer', TfidfVectorizer(stop_words='english', ngram_range=(1, 2))),
        ('classifier', MultinomialNB(alpha=0.1)) ])
```

```python
model.fit(X_train, y_train)
```

```python
def predict_categories(trained_model, dataframe_to_predict):

    df_result = dataframe_to_predict.copy()
    
    # --- 1. Clean the narration column (must be same as training) ---
    narrations_to_predict = df_result['Narration'].fillna('')

    # --- 2. Use the trained model to predict ---
    print("\nPredicting categories...")
    predictions = trained_model.predict(narrations_to_predict)

    # --- 3. Write predictions into the 'account head' column ---
    df_result['Account Head'] = predictions
    print("Predictions have been written to the 'account head' column.")
    
    return df_result
```
